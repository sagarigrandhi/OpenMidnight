```python
import random
from pathlib import Path
```

```python
sample_list_path = "/block/TCGA/sample_dataset_30.txt"
output_25 = "/data/ratna/data_scaling/sample_dataset_30_25percent.txt"
output_75 = "/data/ratna/data_scaling/sample_dataset_30_75percent.txt"
```

```python
samples = []
with open(sample_list_path, "r") as f:
    for line in f:
        parts = line.strip().split()
        if len(parts) == 4:
            svs_path, x, y, level = parts
            samples.append({
                "path": svs_path,
                "x": int(x),
                "y": int(y),
                "level": int(level),
                "line":line
            })

print(f"Loaded {len(samples)} samples from {sample_list_path}")
```

```python
# Get unique slides
slides = sorted(set(s["path"] for s in samples))
print(f"Total unique slides: {len(slides)}")
```

```python
# Sample 25% and 75% at slide level
random.seed(42)
shuffled_slides = slides.copy()
random.shuffle(shuffled_slides)

n_25 = len(shuffled_slides) * 25 // 100
n_75 = len(shuffled_slides) * 75 // 100

slides_25 = set(shuffled_slides[:n_25])
slides_75 = set(shuffled_slides[:n_75])

print(f"25% slides: {len(slides_25)}")
print(f"75% slides: {len(slides_75)}")
```

```python
# Write 25% file
with open(output_25, 'w') as f:
    for s in samples:
        if s["path"] in slides_25:
            f.write(s["line"])
print(f"Written {output_25}")
```

```python
# Write 75% file
with open(output_75, 'w') as f:
    for s in samples:
        if s["path"] in slides_75:
            f.write(s["line"])
print(f"Written {output_75}")
```
