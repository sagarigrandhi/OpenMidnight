```python
import random
```

```python
sample_list_path = "/block/TCGA/sample_dataset_30.txt"
output_25 = "/data/ratna/data_scaling/sample_dataset_30_25percent.txt"
output_50 = "/data/ratna/data_scaling/sample_dataset_30_50percent.txt"
output_75 = "/data/ratna/data_scaling/sample_dataset_30_75percent.txt"
```

```python
samples = []
with open(sample_list_path, "r") as f:
    for line in f:
        parts = line.strip().split()
        if len(parts) == 4:
            svs_path, x, y, level = parts
            samples.append({
                "path": svs_path,
                "x": int(x),
                "y": int(y),
                "level": int(level),
                "line":line
            })

print(f"Loaded {len(samples)} samples from {sample_list_path}")
```

```python
# Get unique slides
slides = sorted(set(s["path"] for s in samples))
print(f"Total unique slides: {len(slides)}")
```

```python
# Sample 25%, 50% and 75% at slide level — independent random draws
n_25 = len(slides) * 25 // 100
n_50 = len(slides) * 50 // 100
n_75 = len(slides) * 75 // 100

random.seed(42)
slides_25 = set(random.sample(slides, n_25))

random.seed(43)
slides_50 = set(random.sample(slides, n_50))

random.seed(44)
slides_75 = set(random.sample(slides, n_75))

print(f"25% slides: {len(slides_25)}")
print(f"50% slides: {len(slides_50)}")
print(f"75% slides: {len(slides_75)}")
```

```python

for pct, slide_set, output_path in [(25, slides_25, output_25), (50, slides_50, output_50), (75, slides_75, output_75)]:
    with open(output_path, 'w') as f:
        for s in samples:
            if s["path"] in slide_set:
                f.write(s["line"])
    print(f"Written {output_path}")
```
