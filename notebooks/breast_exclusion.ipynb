```python
import pandas as pd
```

```python
# Load metadata
metadata_path = "/data/ratna/breast_exclusion/tcga_all_slides.csv"
sample_list_path = "/block/TCGA/sample_dataset_30.txt"
output_path = "/data/ratna/breast_exclusion/sample_dataset_30_no_breast.txt"
```

```python
# Get BRCA slide filenames
tcga_df = pd.read_csv(metadata_path)
brca_slides = set(tcga_df[tcga_df["project_id"] == "TCGA-BRCA"]["slide_file_name"].tolist())
print(len(brca_slides))
```

```python
# Load all lines at once
with open(sample_list_path, "r") as f:
    lines = f.readlines()

original_count = len(lines)
print(f"Original sample list: {original_count}")

# Filter in memory
print("Filtering sample list...")
kept_lines = []
removed = 0
for line in lines:
    slide_name = line.split(" ")[0].split("/")[-1]
    if slide_name in brca_slides:
        removed += 1
    else:
        kept_lines.append(line)

# Write all at once
with open(output_path, "w") as f_out:
    f_out.writelines(kept_lines)

output_count = len(kept_lines)
print(f"Removed: {removed}")
print(f"Output sample list: {output_count}")
print(f"Tally check: {removed + output_count == original_count}")
print("Done")
```

```python
# Sanity check
# Verify no BRCA slides in output
print("Verifying no BRCA slides in output...")
with open(output_path, "r") as f:
    for line in f:
        slide_name = line.split(" ")[0].split("/")[-1]
        if slide_name in brca_slides:
            print(f"BRCA slide found: {slide_name}")
            break
    else:
        print("Verified: No BRCA slides in output file")
```
